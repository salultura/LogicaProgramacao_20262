**Estruturas de repetição**

Imagine que você precisa imprimir na tela os números de 1 a 100. Sem um laço, você teria que escrever print(1), print(2), print(3)... cem vezes! Seria tedioso, ineficiente e propenso a erros. As estruturas de repetição resolvem isso.

**Principais Vantagens:**


**Automação**: Executam a mesma tarefa várias vezes sem a necessidade de reescrever o código.

**Eficiência**: Reduzem drasticamente a quantidade de código, tornando-o mais limpo, legível e fácil de manter.

**Processamento de Dados:** São essenciais para percorrer e processar coleções de dados, como listas de nomes, notas de alunos ou qualquer conjunto de informações.

**Flexibilidade:** Permitem que a repetição seja controlada por uma condição, continuando enquanto for necessário e parando no momento certo.

```python
#Estrutura for com [inicio, fim]
for x in range(1, 11):
  print(x)
```

```python
#Estrutura for com [fim] {inicia do zero}
for a in range(10):
  print(a)
```

```python
#Estrutura for com [inicio, fim, incremento]
for y in range(1, 15, 2):
  print(y)
```

```python
#Estrutura for com [inicio, fim, incremento]
for z in range(20, -5, -3):
  print(z)
```

```python
#Capturar a idade de X pessoas e exibir a media
soma_idades = 0
total_participantes = 10

for x in range(total_participantes):
  idade = int(input('Informe a sua idade: '))
  soma_idades = soma_idades + idade

media = soma_idades/ total_participantes

print('Media de idade:', media)
```

```python
#Imprimir todos os numeros pares entre 10 e 100
for x in range(10, 101):
  if x % 2 == 0:
    print(x)
```

```python
#Imprimir todos os numeros pares entre 10 e 100
for x in range(10, 101, 2):
  print(x)
```

```python
x = 10

while x > 0:
  print(x)
  x = x + 1
```

```python
while True:
  senha = input('Informe a sua senha: ')

  if senha == '1234':
    print('Logado no banco')
    break
  else:
    print('Senha invalida. Tente novamente')
```

```python
senha = ''
while senha != '1234':
  senha = input('Informe a sua senha: ')
  if senha == '1234':
    print('Logado no sistema')
  else:
    print('Senha invalida. Tente novamente')
```

```python
#O continue avança para a proxima repetição
# funciona com while ou for
for x in range(1, 51):
  if x % 5 == 0:
    continue
  print(x)
```

```python
#O break interrompe toda a estrutura de repetição.
# funciona com while ou for
for x in range(1, 51):
  if x % 5 == 0:
    break
  print(x)
```
