```python
#Operadores aritméticos
# adição --> +
# subtração --> -
# multiplicação --> *
# divisão real --> /
# divisão inteira --> //
# resto da divisão --> %
# potenciação --> **
num1 = 10
num2 = 4
print(num1 + num2) #14
print(num1 - num2) #6
print(num1 * num2) #40
print(num1 / num2) #2.5
print(num1 // num2) #2
print(num1 % num2) #2
print(num1 ** num2) #10000
```

```python
#Casting [conversão de tipos]
x = int(input('Informe o valor de X: ')) #convertendo para int
y = int(input('Informe o valor de Y: ')) #convertendo para int

resultado = x * y

print('A multiplicação de X e Y é', resultado)
```

```python
#Crie um programa que calcule a gorjeta de um garçom.
#O programa deve perguntar o valor da despesa e calcular
#10% desse valor.
#Converta esse valor para real.
#Imprima no final o valor da gorjeta.
despesa = float(input('Informe o total da despesa: '))

gorjeta = despesa * 0.1 #10% do valor

print('Gorjeta:', gorjeta)
#Arredondando em 2 casas decimais
print(f'Gorjeta: {gorjeta:.2f}')
```
