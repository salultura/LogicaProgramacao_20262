```python
# Crie um programa que capture 2 notas de um
# aluno e informe a media.

#CASA DECIMAL SEMPRE DEFINIDA COM O CARACTERE '.'
#exemplo: 4.5 --> correto
#exemplo: 7,5 --> erro
n1 = float(input('Informe a nota 01: '))
n2 = float(input('Informe a nota 02: '))

media = (n1 + n2) / 2

print('Media final:', media)

if media >= 7:
  print('Aprovado')
elif media >= 5:
  print('Recuperação')
elif media >= 4:
  print('Prova final')
else:
  print('Reprovado')
```

```python
#Estruturas de decisão

numero = int(input('Informe um numero: '))

if numero % 2 == 0:
  print('PAR')
  print('Fim do programa')
else:
  print('IMPAR')
  print('Encerramento')
```
