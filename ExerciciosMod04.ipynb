```python
#2.1 Crie um programa que solicite ao usuário que
#    digite uma nota entre 0 e 10.
#    Se o usuário digitar um valor inválido
#    (menor que 0 ou maior que 10), exiba "Nota inválida!
#    Digite novamente." e continue repetindo o pedido.
nota = float(input('Informe a nota: '))

while nota < 0 or nota > 10:
  print('Nota inválida! Digite novamente.')
  nota = float(input('Informe a nota: '))
print('Nota registrada com sucesso!')
```

```python
#2.2 Desenvolva um programa que utilize um loop
#    contínuo (while True) para ler números inteiros
#    fornecidos pelo usuário:
#    A cada número digitado, acumule o total da soma.
#    Se o usuário digitar 0, o laço deve ser interrompido
#    imediatamente com o comando break.
#    Ao final, imprima a soma total dos números digitados.
#Acumalador, variaveis 'neutras'
#soma -> 0
#multiplicação -> 1

soma_numeros = 0

while True:
  numero = int(input('Informe um numero: '))
  #soma_numeros = soma_numeros + numero
  soma_numeros += numero

  if numero == 0:
    break

print('Soma dos valores:', soma_numeros)
```

```python
#2.3 Laço for que percorra o intervalo de 1 a 20.
#    Exibindo os valores não são multiplos de 3.
for x in range(1, 21):
  if not x % 3 == 0: #alternativo
    print(x)
```

```python
#2.3 Laço for que percorra o intervalo de 1 a 20.
#    Exibindo os valores não são multiplos de 3.
for x in range(1, 21):
  if x % 3 != 0:
    print(x)
```

```python
#2.4 Suponha que a população do País A seja de 80.000
#    habitantes (com taxa de crescimento anual de 3%)
#    e a do País B seja de 200.000 habitantes
#   (com taxa de 1,5%). Escreva um programa que
#   calcule em quantos anos a população do País A
#   ultrapassará ou igualará a do País B.

anos = 0
pais_A = 80_000
pais_B = 200_000

while pais_B >= pais_A:
  pais_A *= 1.03  #Incrementa 3% do valor atual
  pais_B *= 1.015 #Incrementa 1.5% do valor atual
  anos += 1       #Se passou 1 ano
print('População do pais A:', pais_A)
print('População do pais B:', pais_B)
print('Total de anos para superar:', anos)

```

```python
#3. Crie um programa para registrar os preços das compras de um cliente:
#   Solicite o preço de cada produto repetidamente até que o usuário digite 0
#   ou "fim" para encerrar.
#   Utilize continue caso o usuário informe um valor negativo (preço inválido).
#   Ao finalizar, o programa deve exibir:
#   -A quantidade total de produtos comprados.
#   -O valor total da compra.
#   -O preço médio por produto (formatado com duas casas decimais).
total_comprado = 0
total_produtos = 0
media_produtos = 0

while True:
  preco = float(input('Informe o preço do produto: '))

  if preco < 0:
    #ignora os valores negativos
    continue

  if preco == 0:
    break

  total_comprado += preco
  total_produtos += 1

media_produtos = total_comprado / total_produtos

print('Valor da compra:', total_comprado)
print('Total de produtos:',total_produtos)
print('Preço médio: {:.2f}'.format(media_produtos))
```
