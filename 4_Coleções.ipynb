```python
#Lista --> list []
#Estrutura de dados mutável.
#Armazena dados atraves das posições (index)
#A primeira posição em uma lista é a posição ZERO (0)
lista_numeros = [1,2,4,7,4,8,9,10,2]
print(lista_numeros)

print(lista_numeros[5])
print(lista_numeros[8])
print(lista_numeros[2])

print(lista_numeros[4] * lista_numeros[3])

print(lista_numeros[-2])
```

```python
lista_nomes = [] #lista vazia
#adiciona um elemento no final da lista
lista_nomes.append('Ana')
lista_nomes.append('Carlos')
lista_nomes.append('Juca')
lista_nomes.append('Bruno')
lista_nomes.append('Sara')

print(lista_nomes)
```

```python
#insert --> adiciona um elemento na posição indicada.
lista_nomes.insert(1, 'Juvenal')
print(lista_nomes)
```

```python
#remove --> apaga um elemento da lista pelo valor
lista_nomes.remove('Juca')
print(lista_nomes)
```

```python
#pop --> apaga um elemento da lista pela posição
lista_nomes.pop(3)
print(lista_nomes)
```

```python
#len --> indica o tamanho da lista
print(lista_nomes)
print(len(lista_nomes))
```

```python
#fatiando uma lista [inicio, fim+1]
lista_frutas = ['banana', 'morango', 'laranja',
                'uva', 'melão', 'Pera']
print(lista_frutas[1:4])
```

```python
#Tupla --> tuple ()
#Estrutura de dados imutável.
#Armazena dados atraves das posições (index)
#A primeira posição em uma lista é a posição ZERO (0)

tupla_numeros = (1,2,3,4,5,6)
print(tupla_numeros)
```

```python
senha_secreta = ('SUPER_SECRETO', )
print(senha_secreta[0])
#Uma tupla nao pode ser alterada
#senha_secreta[0] = 'SENHA_FACIL' #ERRO
```

```python
#Conjuntos --> set {}
#Estrutura de dados mutável.
#Nao aceita valores duplicados.
#Os dados não sao organizados atraves de indices.

conjuntos_numeros = {1,2,3,4,5,4}
print(conjuntos_numeros)
```

```python
#Adiciona um elemento ao conjunto
conjuntos_numeros.add(10)
print(conjuntos_numeros)
conjuntos_numeros.add(1)
print(conjuntos_numeros)
```

```python
#pop --> apaga um valor aleatorio do conjunto
print(conjuntos_numeros)
conjuntos_numeros.pop()
print(conjuntos_numeros)
```

```python
#Pessoa
nome = 'Juca'
idade = 23
profissao = 'Engenheiro'
salario = 8000
```

```python
#Dicionários --> dict {key : value}
#Armazena os valores no formato chave : valor
pessoa = {
    'nome' : 'Juca',
    'idade' : 23,
    'profissao' : 'Engenheiro',
    'salario' : 8000
}

#Imprimindo o dicionario
print(pessoa)
print(pessoa['nome'])
print(pessoa['idade'])
print(pessoa['profissao'])
print(pessoa['salario'])
```

```python
#Adicionando uma nova chave ao dicionario
pessoa['sexo'] = 'Masculino'
print(pessoa)
```

```python
#Removendo uma chave de um dicionario
del(pessoa['idade'])
print(pessoa)
```

# Exercícios

Crie uma lista vazia chamada tarefas e execute sequencialmente os passos abaixo:

* Adicione as tarefas "Estudar Python" e "Lavar a louça" ao final da lista usando .append().

* Insira a tarefa "Arrumar a cama" no início da lista (posição 0) usando .insert().

* Remova a tarefa "Lavar a louça" usando .remove().


* Remova o primeiro elemento da lista utilizando .pop(0) e guarde-o em uma variável chamada tarefa_concluida.

* Exiba na tela a tarefa_concluida e a lista tarefas restante.


```python
tarefas = []
tarefas.append('Estudar Python')
tarefas.append('Lavar a louça')
print(tarefas)
tarefas.insert(0, 'Arrumar a cama')
print(tarefas)
tarefas.remove('Lavar a louça')
print(tarefas)
tarefa_concluida = tarefas.pop(0)
print(tarefas)
print(tarefa_concluida)
```

Crie um dicionário chamado produto contendo as seguintes chaves e valores:

* "nome": "Notebook"

* "preco": 3500.00

* "estoque": 15

Em seguida:

* Atualize o preço para 3200.00.
* Adicione uma nova chave "categoria" com o valor "Eletrônicos".
* Remova a chave "estoque" usando .pop().
* Percorra o dicionário usando um laço for junto de .items() e imprima cada chave e seu respectivo valor.


```python
produto = {'nome' : 'Notebook', 'preco' : 3500,
           'estoque' : 15
}
print(produto)
produto['preco'] = 3200
print(produto)
produto['categoria'] = 'Eletrônicos'
print(produto)
produto.pop('estoque')
print(produto)

for item in produto.items():
  print(item)
```
