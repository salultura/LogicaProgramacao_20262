2. Criando Funções em Python
A sintaxe para criar uma função em Python é simples e utiliza a palavra-chave def.
Estrutura:




```python
def nome_da_funcao(parametros):
    # Bloco de código da função (indentado)
    # Este código será executado quando a função for chamada
    pass # A palavra 'pass' é usada quando não queremos adicionar lógica ainda
```

Definir uma função não a executa. É como escrever uma receita e guardá-la no livro. Para que o código dentro da função seja executado, você precisa chamá-la. Para chamar uma função, basta escrever seu nome seguido por parênteses.


```python
#Criando uma função em parametros
def mensagem():
  print('Bem-vindos')
  print('Hoje teremos aula de python')
  print('dia 20/11 será feriado')
  print('tchau...')

#chamando/executando a função
mensagem()
mensagem()
mensagem()
```

```python
#função com parâmetros
# Crie um programa que receba o nome e o sobrenome de um funcionario
# a função deve imprimir o email institucional
#carlos antonio ==> carlos.antonio@email.com

def gera_email(nome, sobrenome):
  print(nome + '.' + sobrenome + '@email.com')

gera_email('juca', 'pereira')
gera_email('luciana', 'goncalves')
gera_email('jose', 'mariano')

```

```python
#Função com retorno
def gere_email_com_retorno(nome, sobrenome):
  return nome + '.' + sobrenome + '@email.com'

gere_email_com_retorno('juca', 'pereira')
gere_email_com_retorno('luciana', 'goncalves')
gere_email_com_retorno('jose', 'mariano')

def converte_nomes_lista(nome1, nome2, nome3):
  return [nome1, nome2, nome3]

nomes = converte_nomes_lista('Paulo', 'Otavio', 'Marcus')

print(nomes)
```
