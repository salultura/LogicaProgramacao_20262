1- Escreva um programa que solicite ao usuário uma temperatura em graus Celsius. O programa deve calcular a conversão para Fahrenheit usando a fórmula F = (C * 9/5) + 32. Em seguida, avalie o resultado: se a temperatura em Fahrenheit for maior que 100, exiba a mensagem "Alerta: Temperatura extrema!". Caso contrário, exiba apenas o valor convertido e a mensagem "Temperatura dentro do normal.".

```python
temp_c = float(input('Temperatura em Celsius: '))

temp_f = (temp_c * 9 / 5) + 32

if temp_f > 100:
  print('Alerta: Temperatura Extrema!', temp_f, 'F°')
else:
  print('Temperatura dentro do normal.', temp_f,'F°')
```

2- Crie um programa para auxiliar no caixa de uma loja. O programa deve pedir ao usuário que digite o valor total da conta e o valor em dinheiro entregue pelo cliente. Se o valor entregue for menor que o valor da conta, exiba a mensagem "Pagamento insuficiente" e mostre quanto dinheiro ainda falta. Caso contrário, calcule e exiba a mensagem "Troco a devolver: R$ " seguido do valor exato do troco.


```python
total_conta = float(input('Total da conta: '))
valor_recebido = float(input('Valor recebido: '))

troco = valor_recebido - total_conta

if troco >= 0:
  print('Troco a develover R$', troco)
else:
  print('Pagamento insuficiente. Falta R$', troco * -1)
```

3- Faça um programa que peça ao usuário para digitar dois números inteiros. O programa deve comparar os valores e imprimir qual é o maior. Se ambos forem idênticos, o programa deve exibir a mensagem "Os números são iguais". (Dica: utilize a estrutura if, elif e else para cobrir todos os cenários sem precisar combiná-los).

```python
n1 = int(input('Informe o numero 01: '))
n2 = int(input('Informe o numero 02: '))

if n1 > n2:
  print('Maior numero:', n1)
elif n2 > n1:
  print('Maior numero:', n2)
else:
  print('Numeros iguais.')
```

4- Desenvolva um programa que leia o valor total de uma compra. Se o valor informado for estritamente maior que R$150,00, o programa deve calcular e aplicar um desconto de 15% sobre o total. Ao final, independentemente de ter recebido desconto ou não, exiba o valor final a ser pago.


```python
total_compra = float(input('Valor da compra: '))

if total_compra > 150:
  desconto = total_compra * 0.15
  valor_final = total_compra - desconto
  print('Valor a pagar: R$', valor_final)
else:
  print('Valor a pagar: R$', total_compra)
```

5- Escreva um programa que solicite um valor numérico ao usuário (pode ser decimal). O sistema deve avaliar a entrada e imprimir no console se o número é "Positivo", "Negativo" ou "Zero", utilizando apenas verificações sequenciais ou aninhadas.

```python
numero = int(input('Informe um numero: '))

if numero > 0:
  print('O numero informado é Positivo.')
elif numero < 0:
  print('O numero informado é Negativo.')
else:
  print('O numero informado é Zero.')
```

6- Crie um programa que leia a velocidade atual de um carro em km/h. Se a velocidade ultrapassar o limite de 80 km/h, o programa deve exibir a mensagem "Você foi multado!" e calcular o valor da multa. A multa custa R$7,00 para cada quilômetro excedente acima do limite. Exiba o valor total da infração. Caso a velocidade seja menor ou igual a 80 km/h, exiba apenas a mensagem "Boa viagem!".

```python
vel_atual = int(input('Velocidade atual: '))

if vel_atual > 80:
  multa = ((vel_atual -80) * 7)
  print('Voce foi multado! Multa a pagar R$', multa)
else:
  print('Boa viagem!')
```
