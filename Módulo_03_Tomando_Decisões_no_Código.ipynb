1- Determine o resultado (True ou False) de cada uma das expressões lógicas e relacionais abaixo, considerando x = 10, y = 20 e z = 10:

x == z

x != y

x > y

(x <= z) and (y > x)

(x > 15) or (z == 10)

not (x == z)

not (y < x) and (x == 10)



```python
#Variavel
x = 10
y = 20
z = 10

print(x == z) #True
print(x != y) #True
print(x > y)  #False
print((x <= z) and (y > x)) #True
print(not (x == z)) #False
print(not (y < x) and (x == 10)) #True
print((x > 15) or (z == 10)) #True
```

2.2: Desenvolva um programa que receba três medidas inteiras para os lados A, B e C:

1- Primeiro, verifique se os lados podem formar um triângulo (a soma de dois lados quaisquer deve ser sempre maior que o terceiro lado).

2- Se formarem um triângulo válido, classifique-o usando if/elif/else:

* Equilátero: Três lados iguais.

* Isósceles: Dois lados iguais.

* Escaleno: Todos os lados diferentes (!=).

Se não formarem um triângulo, exiba "Medidas inválidas para formar um triângulo.".


```python
lado_a = int(input('Informe o lado A: '))
lado_b = int(input('Informe o lado B: '))
lado_c = int(input('Informe o lado C: '))

if lado_a + lado_b > lado_c and \
   lado_a + lado_c > lado_b and \
   lado_b + lado_c > lado_a:
   #triangulo valido
  if lado_a == lado_b and lado_a == lado_c:
    print('Triangulo equilatero')
  elif lado_a == lado_b or lado_a == lado_c:
    print('Triangulo isosceles')
  else:
    print('Triangulo escaleno')
else:
  print('Medidas inválidas para formar um triângulo."')
```

2.3: Crie um programa que leia o peso (kg) e a altura (m) de uma pessoa, calcule o IMC [IMC=peso/(altura * altura)] e classifique o resultado utilizando if/elif/else:

* IMC <18.5: "Abaixo do peso"
* 18.5 <= IMC < 25.0: "Peso ideal"
* 25.0 <= IMC < 30.0: "Sobrepeso"
* IMC >= 30.0: "Obesidade"


```python
peso = float(input('Informe o seu peso: '))
altura = float(input('Informe a sua altura: '))

imc = peso / altura ** 2

print('Seu IMC:', imc)
print(f'Seu IMC: {imc:.3f}')

if imc < 18.5:
  print('Abaixo do peso.')
elif imc >= 18.5 and imc < 25:
  print('Peso ideal.')
elif imc >= 25 and imc < 30:
  print('Sobrepeso.')
else:
  print('Obesidade.')
```

2.4 Crie um programa que classifica o nível de um poluente no ar com base em sua medição em microgramas por metro cúbico (µg/m³).

O programa deve seguir a seguinte tabela:

Medição (µg/m³) / Classificação

0 a 50 => Boa

51 a 100 => Moderada

101 a 150 => Ruim

Acima de 150 => Crítica



```python
medicao = float(input('Informe o valor medido: '))

if medicao >= 0 and medicao <= 51:
  print('Boa')
elif medicao > 51 and medicao <= 101:
  print('Moderada')
elif medicao > 101 and medicao <= 150:
  print('Ruim')
elif medicao > 151:
  print('Critica')
else:
  print('Medicao invalida')
```
