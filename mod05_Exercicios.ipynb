1.1 Crie uma lista vazia chamada tarefas e execute sequencialmente os passos abaixo:

1.2 Dada a lista numeros = [45, 12, 89, 3, 27, 64, 18]:

* Use fatiamento (slicing) para extrair os três primeiros elementos em uma nova lista.

* Use fatiamento para extrair os elementos do índice 3 até o final.

* Ordene a lista original numeros em ordem crescente utilizando .sort().


* Verifique se o número 27 está contido na lista usando o operador in e exiba o resultado booleano.


```python
numeros = [45, 12, 89, 3, 27, 64, 18]
tres_primeiros = numeros[0:3]
outros_numeros = numeros[3:7]

print(numeros)
print(tres_primeiros)
print(outros_numeros)
numeros.sort()#Ordem crescente
print(numeros)
numeros.sort(reverse=True)#Ordem decrescente
print(numeros)
print(27 in numeros)
```

2.1
Crie um dicionário chamado produto contendo as seguintes chaves e valores:

* "nome": "Notebook"

* "preco": 3500.00

* "estoque": 15

Em seguida:

* Atualize o preço para 3200.00.
* Adicione uma nova chave "categoria" com o valor "Eletrônicos".
* Remova a chave "estoque" usando .pop().
* Percorra o dicionário usando um laço for junto de .items() e imprima cada chave e seu respectivo valor.


```python
produto = {'nome' : 'Notebook', 'preco' : 3500,
           'estoque' : 15
}
print(produto)
produto['preco'] = 3200
print(produto)
produto['categoria'] = 'Eletrônicos'
print(produto)
produto.pop('estoque')
print(produto)

for item in produto.items():
  print(item)
```

2.2 Dado o dicionário usuario = {"nome": "Beatriz", "perfil": "Admin"}:


* Tente acessar as chaves "nome" e "email" utilizando o método .get().

* Para o caso de "email", defina o valor padrão "Não informado" caso a chave não exista.


```python
usuario = {"nome": "Beatriz", "perfil": "Admin"}
usuario.setdefault('email', 'Não informado')

print(usuario.get('nome'))
print(usuario.get('email'))
```

3.2 Considere as duas turmas de oficinas com os seguintes participantes matriculados:

oficina_python = {"Carlos", "Mariana", "Lucas", "Ana"}

oficina_web = {"Ana", "Beatriz", "Lucas", "Rodrigo"}


Escreva as expressões para obter:

* Todos os participantes únicos inscritos em pelo menos uma oficina (União: |).
* Os participantes que estão inscritos em ambas as oficinas (Interseção: &).
* Os participantes inscritos apenas na oficina de Python (Diferença: -).


```python
oficina_python = {"Carlos", "Mariana", "Lucas", "Ana"}
oficina_web = {"Ana", "Beatriz", "Lucas", "Rodrigo"}

todos_participantes = oficina_python | oficina_web #união
ambas_oficinas = oficina_python & oficina_web #interseção
apenas_python = oficina_python - oficina_web #diferença

print(todos_participantes)
print(ambas_oficinas)
print(apenas_python)
```

##Parte 4: Desafio Integrador e Coleções Aninhadas

4.1 Crie uma lista chamada turma contendo 3 dicionários, onde cada dicionário representa um aluno com "nome" (str) e "notas" (uma lista de 3 números decimais).
Percorra a lista turma com um laço for.
Para cada aluno, calcule a média de suas notas.
Exiba o nome do aluno e sua respectiva média calculada com 1 casa decimal.


```python
turma = [
    {'nome':'Juca', 'notas' : [7, 6, 6]},
    {'nome' : 'Beatriz', 'notas' : [5, 8, 7]},
    {'nome' : 'João', 'notas' : [10, 9, 8]}
]

for aluno in turma:
  media = sum(aluno['notas']) / 3
  print('Aluno:', aluno['nome'],
        '| Media:{:.1f}'.format(media) )
```
